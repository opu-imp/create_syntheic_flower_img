```python
import cv2
import numpy as np

def extract_flower_contour(image_path):
    """
    花の輪郭を抽出する関数
    
    Args:
        image_path (str): 入力画像のパス
        
    Returns:
        tuple: (マスク画像, 切り出した花画像)。画像読み込みに失敗した場合はNoneを返す
    """
    # 画像を読み込む
    image = cv2.imread(image_path)
    if image is None:
        print("画像が読み込めません")
        return None
        
    # 画像をグレースケールに変換
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    
    # 画像をぼかす
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)
    
    # 二値化
    _, thresh = cv2.threshold(blurred, 150, 255, cv2.THRESH_BINARY_INV)
    
    # 輪郭を検出
    contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    
    # 輪郭の中から最大のものを選択（花の領域と仮定）
    if not contours:
        print("輪郭が検出できませんでした。")
        return None
        
    largest_contour = max(contours, key=cv2.contourArea)
    
    # マスクを作成
    mask = np.zeros_like(image)
    cv2.drawContours(mask, [largest_contour], -1, (255, 255, 255), thickness=cv2.FILLED)
    
    # 切り出した画像を取得
    flower_only = cv2.bitwise_and(image, mask)
    
    return mask, flower_only
```

```python
import json

# labeled/bb_point/から実画像のアノテーションを読み込む
def get_ground_truth(filename):
    # ファイル名を取得
    file_name = os.path.basename(filename)
    json_path = f'labeled/bb_point/{file_name.replace(".png", ".json")}'
    
    # JSONファイルを読み込む
    with open(json_path, 'r') as f:
        data = json.load(f)
    
    # keypointのアノテーションを取得
    keypoints = []
    classes = []
    for ann in data['annotations']:
        if ann['type'] == 'keypoint':
            keypoints.append(ann['points'])
            # attributesからpointの値を取得
            point_value = int(ann['attributes'][0]['value'])
            classes.append(point_value)
    
    # keypointをbboxの形式に変換 [x, y, width, height]
    bboxes = [[p[0], p[1], 10, 10] for p in keypoints]
    
    return classes, bboxes
```
